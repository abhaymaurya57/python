```python
def person_lister(f):
    def inner(people):
        people.sort(key=lambda person: int(person[2]))

        return [f(person) for person in people]

    return inner


@person_lister
def name_format(person):
    return ("Mr. " if person[3] == "M" else "Ms. ") + person[0] + " " + person[1]


if __name__ == '__main__':
    people = [
        ['Mike', 'Thomson', '20', 'M'],
        ['Robert', 'Bustle', '32', 'M'],
        ['Andria', 'Bustle', '30', 'F']
    ]
    print(*name_format(people), sep='\n')
```

```python
total = 0
for employee in employees:
    if employee['department']=="IT" and employee['salary']>50000:
        total+=employee['salary']
print(total)
```

```python
dic={}
for employee in employees:
    dep = employee['department']
    if dep not in dic:
        dic[dep]  = employee['salary']
    else:
        dic[dep] += employee['salary']
print(dic)
```
