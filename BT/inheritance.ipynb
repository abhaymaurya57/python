```python
class MySQL:

    def connect(self):
        print(f'MySQL database connect....')

class Postgres:

    def connect(self):
        print(f'Postgres database connected....')

class DatabaseFactory:

    @staticmethod
    def connectdb(dbname):

        if dbname == 'mysql':
            return MySQL()

        if dbname == 'postgres':
            return Postgres()
        else:
            raise ValueError("Database name is not match")

db = DatabaseFactory.connectdb('mysql')
db.connect()
```

```python
triangle =[]

def pascal(n):
    for i  in range(n):
        if i <= 0:
           triangle.append([1])
        else:
            prev = triangle[:-1]
            newlst = []
            newlst.append(1)
            for j in range(1, len(prev)):
               newlst.append(prev[j]+prev[j-1])
            newlst.append(1)
            triangle.append(newlst)

pascal(5)
for i in triangle:
    print(i)
```

```python
class Book:

    def __init__(self,name,age):
        self.name = name
        self.age = age

    def __str__(self):
        print(f'This is  str')
        return self.name

    def __len__(self):
        print(f'Name length is {len(self.name)}')
        return len(self.name)


    def __setattr__(self, name, value):
        if isinstance(value,str)  and len(value)<2:
            raise ValueError("lenght is less than 2")
        return super().__setattr__(name, value)

    def __delattr__(self,value):
        print(f"Deleteing {value}")
        super().__delattr__(value)


obj = Book('32',2)
print(obj)
print(dir(obj))
print(obj.__class__)
print(len(obj))
print(obj.name)
print(obj.age)
del obj.name
# print(obj.name)
print(obj.age)
print(repr(obj))
```

```python
class Counter:

    def __init__(self):
        self.i = 0

    def __iter__(self):
        return self

    def __next__(self):
        self.i += 1
        if self.i > 3:
            raise StopIteration
        return self.i

for x in Counter():
    print(x)
```

```python
class Number:

    def __init__(self, x):
        print('init called ',x)

        self.x = x

    def __add__(self, other):
        print(f'add called',self.x)
        print(f'add called',other.x)
        return self.x + other.x


a = Number(5)
b = Number(10)
# c = Number(10)

print(a + b )
```

```python
class Num:

    def __init__(self,x):
        self.x = x

    def __mul__(self, other):
        print("multiply is called")
        return self.x * other.x
    
a = Num(3)
b = Num(4)
print(a*b)
        
```
