```python
from abc import ABC, abstractmethod

class Animal(ABC):
    def __init__(self, name):
        self.name = name

    @abstractmethod
    def sound(self):
        pass

class Dog(Animal):
    def sound(self):
        print(f"{self.name} says Bark")

dog = Dog("dog")
dog.sound()
```

```python
class Engine:

    def __init__(self):
        print("Engine Created")

    def engine_start(self):
        print("Engine Started")

    def engine_run_speed(self):
        print("Speed = 100 km/h")


class Car:

    _name = "Abhay"

    def __new__(cls):
        cls._name += " Maurya"
        return super().__new__(cls)

    def __init__(self):
        print("Car Created")
        self.engine = Engine()

    def car_start(self):
        self.engine.engine_start()
        self.engine.engine_run_speed()
        print("Car Started")


car = Car()

print(Car._name)

car.car_start()
```

```python
class Engine:

    def __init__(self):
        print(f'This is Engine class')

    @property
    def engine_start(self):
        print(f'finally engine is started....')

    def engine_speed(self):
        print(f'engine Speed is 100km/s')

class Car:

    def __init__(self,engine):
        self.engine = engine

    def car_start(self):
        self.engine.engine_start
        self.engine.engine_speed()
        print(f'car is started....')

engine = Engine()
car = Car(engine)
car.car_start()
```

```python
class Teacher:

    def __init__(self, name):
        self.name = name


class Student:

    def __init__(self, teacher, name):
        self.teacher = teacher
        self.name = name

    def learn(self):
        print(f"{self.name} learns from teacher {self.teacher.name}")


teacher = Teacher("Vinay")
student = Student(teacher, "Abhay")

student.learn()
```

```python
class Count:
    def __init__(self):
        self.num = 1

    def __iter__(self):
        return self

    def __next__(self):
        if self.num > 10:
            raise StopIteration

        value = self.num
        self.num += 1
        return value


obj = Count()

for i in obj:
    print(i)
```
