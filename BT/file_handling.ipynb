```python
import csv

with open("employee.csv","w",newline="") as f:

    writer=csv.writer(f)

    writer.writerow(["Name","Age"])
    writer.writerow(["Abhay",22])
    writer.writerow(["Rahul",23])
```

```python
rows = [
    ["Name", "Age"],
    ["Abhay", 22],
    ["Rahul", 23],
    ["Amit", 24],
    ["Name", "Age"],
    ["Abhay", 22],
    ["Rahul", 23],
    ["Amit", 24],
]

import csv

with open("employee.csv", "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerows(rows)
```

```python
import csv

with open("employee.csv","w",newline="\n") as f:
    fieldnames = ["Name","Age"]

    writer = csv.DictWriter(f,fieldnames=fieldnames)

    writer.writeheader()

    writer.writerow({
        "Name":"Abhay",
        "Age":22
    })

    writer.writerow({
        "Name":"Rahul",
        "Age":23
    })
```

```python
import csv
from datetime import date

with open("attendance.csv", "a", newline="") as f:
    writer = csv.writer(f)
    print(writer.writerow([date.today(), "Abhay", "Present"]))
```

```python
import csv

rows = [
    {"Name": "Abhay", "Age": 22, "Marks": 95},
    {"Name": "Rahul", "Age": 23, "Marks": 88},
]

with open('std.csv','w',newline="") as f:
    writer = csv.DictWriter(
        f,
        fieldnames=["Name","Age","Marks"]
    )

    writer.writeheader()
    writer.writerows(rows)
```
