```python
def decorator(func):

    def wrapper():
        print("Before")
        func()
        print("after")

    return wrapper

@decorator
def hello():
    print("Hello")

print(hello.__name__)
print(hello())
```

```python
import inspect

def log(func):

    def wrapper(*args, **kwargs):

        print(inspect.signature(func))

        return func(*args, **kwargs)

    return wrapper

@log
def add(a: int, b: int):
    return a + b

print(add(2, 3))
```

```python
class Student:
    _name = 'abhay'
    print("r")

    def __call__(self, *args, **kwds):
        print(f'call')

    def __new__(cls):
        print(f'new {cls._name}')

    def model(self):
        print(f'model {self._name}')

obj = Student()
print(callable(obj))
print(obj.model())


# print(callable(obj))
```
