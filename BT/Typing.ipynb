#Typing

```python
from typing import Dict,Any
```

```python
score : Dict[str , str] = {}
score["Name"] = "abhay"
print(score)
```

```python
score : Dict[str,Any] = {}
score["name"] = 21
print(score)

score[21] = "age"
print(score)

score['Abhay',12] = "da"
print(score)

```

```python
dic = {}
dic['name'] = 'abhay'
print(dic)
```

```python
dic['abhay' , 21] = "age"
print(dic)

print(dic['abhay',21])
```

```python
score : dict[str,str] = {}
score[20]  = 40
print(score)
```

```python
from functools import wraps

def my_decorator(func):
    # wraps copies metadata from func to wrapper
    @wraps(func)
    def wrapper(*args, **kwargs):
        print("Something is happening before the function is called.")
        result = func(*args, **kwargs)
        print("Something is happening after the function is called.")
        return result
    return wrapper

@my_decorator
def say_hello(name: str) -> str:
    """Greets a user."""
    return f"Hello, {name}!"

# Without @wraps, this would print 'wrapper' instead of 'say_hello'
print(say_hello.__name__) 

# Preserves the original docstring
print(say_hello.__doc__) 

# Preserves type annotations
print(say_hello.__annotations__)
```

```python
import textwrap

def wrap(string, max_width):
    return textwrap.fill(string, max_width)

if __name__ == '__main__':
    string = input()
    max_width = int(input())
    result = wrap(string, max_width)
    print(result)
```

```python
def capital(st):
    lst = st.split(" ")
    print(lst)
    newlst = []
    for i in lst:
        ch =  i[0]
        print(ch)
        print(chr(ord(ch)-32))
        
        newlst.append(chr(ord(ch)-32)+i[1:])
    print(newlst)


capital("dada fwfds rewqr")
```
