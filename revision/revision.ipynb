```python
from abc import ABC,abstractmethod

class Models(ABC):

    @abstractmethod
    def model():
        pass

class Table(Models):

    def model(self):
        print(f'this is my model')

class Dog(Models):
    def model():
        print(f'this is Dog class')

class Book(Models):
    pass


obj = Table()
obj.model()

Dog.model()
Book()
```
