```python
vowel = ['a','e','i','o','u']
word='programming'
count=0
for i in word:
    if i in vowel:
        count+=1
print(count)
```

```python
vowel = ['a','e','i','o','u']
word='programming'
count=0
for i in word:
    if i not in vowel:
        count+=1
print(count)
```

```python
fib = [0,1]

n=5
for i in range(n):
    fib.append(fib[-1]+ fib[-2])
    print(fib)
print(', '.join(str(e) for e in fib))
```
