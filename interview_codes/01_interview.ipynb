```python
def pyramid(n):
    for i in range(n):
        for j in range(n-i):
            print(" ",end="")
        for j in range(i+1):
            print("*",end="")
        for j in range(i):
            print("*",end="")
        print("")

pyramid(5)
```

```python
def is_prime(num):
    for i in range(2,num):
        if num%i==0:
            return False
    return True

def prime_generator(n):
    num=2
    while n:
        if is_prime(num):
            yield num
            n-=1
        num+=1

x  =  int(input("Entr no. of prime numbers required"))
obj = prime_generator(x)
for i in  obj:
    print(i,end=" ")
```

```python
# decorator

def fun(func):
    def decorator(*args,**kwargs):
        print(f'this is decorator')
        res = func(*args,**kwargs)
        print(f'func return value is: ',res)
        print(f'After decorator')
    return decorator

@fun
def func(a,b):
    print(f'this is add function')
    return a+b

func(4,3)
```
