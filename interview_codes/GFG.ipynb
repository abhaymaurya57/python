```python
def add(x,y):
    return x+y

def apply_func(fun,a,b):
    return fun(a,b)

print(apply_func(add,a=3,b=4))
print(apply_func(add,4,5))
```

```python
def call_by_val(x):
    x = x*2
    print(f'x : {id(x)}')
    return x

def call_by_ref(b):
    b.append("D")
    print(f'b: {id(b)}')
    return b

a = ["E"]
num = 6
print(f'num: {id(num)}')
print(f'a : ',id(a))
#call functions
update_num = call_by_val(num)
update_list = call_by_ref(a)

# print after function calls
print("Updated value after call_by_val: ",update_num)
print("Updated list after call_by_ref:",update_list)
```
