```python
# correct solution 
# Chat-GPT solution

def reverse_vowels(s):
    vowels = "aeiouAEIOU"
    arr = list(s)
    left = 0
    right = len(arr) - 1
    while left < right:
        # move left until vowel
        while left < right and arr[left] not in vowels:
            left += 1
        # move right until vowel
        while left < right and arr[right] not in vowels:
            right -= 1
        # swap vowels
        arr[left], arr[right] = arr[right], arr[left]
        left += 1
        right -= 1
    return "".join(arr)
# test
print(reverse_vowels("hello"))   # holle
print(reverse_vowels("abhay"))   # abhay
```
